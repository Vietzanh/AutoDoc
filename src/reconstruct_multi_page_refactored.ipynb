# Multi-Page PDF to DOCX Reconstruction (Refactored)

This notebook reconstructs a DOCX file from a multi-page PDF using the refactored pipeline.

The code has been refactored into modular components:
- `pipeline.py`: Main pipeline class
- `model_loader.py`: Model loading utilities
- `docx_generator/`: DOCX generation logic
- `utils/`: Utility functions


```python
import sys
from pathlib import Path

# Add parent directory to path so we can import src modules
current_dir = Path.cwd()
if current_dir.name == 'src':
    # If we're in src directory, go up one level
    parent_dir = current_dir.parent
    sys.path.insert(0, str(parent_dir))
else:
    # Otherwise, add current directory
    sys.path.insert(0, str(current_dir))

from src.pipeline import PDFToDocxPipeline
```

```python
# Configuration
PDF_PATH = "../examples/Test.pdf"
OUTPUT_PATH = "../docx_file/reconstructed_multi_page_refactored.docx"
JSON_BASE_PATH = "../data_layout"
START_PAGE = 0
END_PAGE = None  # None = all pages
```

```python
# Initialize pipeline
pipeline = PDFToDocxPipeline(
    model=None,  # Will be loaded automatically
    style_map=None,  # Will use default
    max_image_width=6.0,  # inches
    dpi=300,
)
```

```python
# Process PDF and generate DOCX
pipeline.process_pdf(
    pdf_path=PDF_PATH,
    output_path=OUTPUT_PATH,
    json_base_path=JSON_BASE_PATH,
    start_page=START_PAGE,
    end_page=END_PAGE,
)
```
